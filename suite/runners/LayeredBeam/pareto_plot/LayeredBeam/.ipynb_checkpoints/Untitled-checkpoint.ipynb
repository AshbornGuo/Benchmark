```python
import numpy as np
import pandas as pd
import ast
import matplotlib.pyplot as plt
from pymoo.util.nds.non_dominated_sorting import NonDominatedSorting
from pathlib import Path

def read_res_raw(path_csv):
    df = pd.read_csv(path_csv, sep=";")
    F = np.vstack(df["objectives"].map(ast.literal_eval).to_numpy())
    return F[:, :2]  # 只取前2个目标

def plot_first_k_fronts_line(F, k=5, save_path=None):
    F = np.asarray(F)
    nds = NonDominatedSorting()
    fronts = nds.do(F)[:k]

    plt.figure(figsize=(7, 6))

    for i, idx in enumerate(fronts):
        pts = F[idx]

        # 为了连线像“前沿”，按 f1 排序再连
        order = np.argsort(pts[:, 0])
        pts = pts[order]

        plt.plot(pts[:, 0], pts[:, 1], marker="o", linewidth=1.5, markersize=3,
                 label=f"Front {i} (n={len(idx)})")

    plt.xlabel("f1 (raw)")
    plt.ylabel("f2 (raw)")
    plt.title(f"Seed331: First {k} Pareto Front Layers (line)")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()

    if save_path is not None:
        save_path = Path(save_path)
        save_path.parent.mkdir(parents=True, exist_ok=True)
        plt.savefig(save_path, dpi=300, bbox_inches="tight")

    plt.show()  # ⭐ 关键：直接弹出图，不是只保存
    plt.close()

    return fronts

# ====== RUN (seed331) ======
PATH_qLogNEHVI_331 = "LayeredBeam_qLogNEHVI_seed331.csv"
F_raw_331 = read_res_raw(PATH_qLogNEHVI_331)

fronts_5 = plot_first_k_fronts_line(
    F_raw_331,
    k=10,
    save_path="qLogNEHVI_seed331_raw_line.png"
)

for i, idx in enumerate(fronts_5):
    print(f"front {i}: {len(idx)} points")
```
